```python
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
import re

# Download stopwords if not already done (run once)
nltk.download('stopwords')

# Initialize stemmer
ps = PorterStemmer()

reviews = [
   "The boy is good",
   "the girl is good",
   "boy and girl are good"
]

stop_words = set(stopwords.words('english'))
corpus = []

for i in range(len(reviews)):
    review = re.sub('[^a-zA-Z]', ' ', reviews[i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stop_words]
    review = ' '.join(review)
    corpus.append(review)


```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(max_features=12,ngram_range=(1,2))
x=tfidf.fit_transform(corpus).toarray()
print(x)
print("Corpus after preprocessing:")
print(corpus)
print("\nFeature names (vocabulary):")
print(tfidf.get_feature_names_out())
print("\nTF-IDF matrix shape:", X.shape)
print("\nTF-IDF matrix:")
print(X)
```
