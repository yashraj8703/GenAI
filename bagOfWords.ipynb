```python
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
import re
from sklearn.feature_extraction.text import CountVectorizer

# Download stopwords if not already done (run once)
nltk.download('stopwords')

# Initialize stemmer
ps = PorterStemmer()

reviews = [
    "I absolutely loved the movie! It was fantastic, thrilling, and full of surprises.",
    "The movie was terrible. It was boring and too long.",
    "Amazing acting and great story, but the movie was a bit slow in the middle.",
    "I didn't like the movie; it was dull and predictable.",
]

stop_words = set(stopwords.words('english'))
corpus = []

for i in range(len(reviews)):
    review = re.sub('[^a-zA-Z]', ' ', reviews[i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stop_words]
    review = ' '.join(review)
    corpus.append(review)

cv = CountVectorizer(max_features=12, binary=True)
X = cv.fit_transform(corpus).toarray()

print("Corpus after preprocessing:")
print(corpus)
print("\nFeature names (vocabulary):")
print(cv.get_feature_names_out())
print("\nBag of Words matrix shape:", X.shape)
print("\nBag of Words matrix (binary):")
print(X)

cv.vocabulary_
sorted_vocab = dict(sorted(cv.vocabulary_.items(), key=lambda x: x[0], reverse=True))
sorted_vocab
```

```python
import numpy as np

# Sum the binary presence across all documents to get word frequencies
word_freq = X.sum(axis=0)  # sums column-wise counts

# Get the feature (word) names
feature_names = cv.get_feature_names_out()

# Get indices of top 5 words by frequency (descending order)
top5_indices = np.argsort(word_freq)[::-1][:5]

print("Top 5 words across all reviews and their frequencies:")
for idx in top5_indices:
    print(f"{feature_names[idx]}: {word_freq[idx]}")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
import re
from sklearn.feature_extraction.text import CountVectorizer

# Download stopwords if not already done (run once)
nltk.download('stopwords')

# Initialize stemmer
ps = PorterStemmer()

reviews = [
   "The Food is Good",
   "The Food is Not negative Good"
]

stop_words = set(stopwords.words('english'))
corpus = []

for i in range(len(reviews)):
    review = re.sub('[^a-zA-Z]', ' ', reviews[i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stop_words]
    review = ' '.join(review)
    corpus.append(review)

cv = CountVectorizer(max_features=100, binary=True,ngram_range=(3,3))
X = cv.fit_transform(corpus).toarray()

print("Corpus after preprocessing:")
print(corpus)
print("\nFeature names (vocabulary):")
print(cv.get_feature_names_out())
print("\nBag of Words matrix shape:", X.shape)
print("\nBag of Words matrix (binary):")
print(X)

cv.vocabulary_
# sorted_vocab = dict(sorted(cv.vocabulary_.items(), key=lambda x: x[0], reverse=True))
# sorted_vocab
```
